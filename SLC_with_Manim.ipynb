A Clique refers to the set of Vertices within a Graph, such that every Vertex is connected to every other Vertex wihin that Clique-set through an Edge.

# Scene 3: The Maximum Clique Problem (MCP)

## Speak/Voice:

The Maximum Clique Problem of the MCP strives to find the largest sized clique within a graph. This means that any MCP algorithm must give the output as the set of Vertices of the Graph which form a Clique, and ensure that no other Clique of that Graph has more number of vertices.

A Graph may have more than one Maximum Clique.

As an example, in the example graph provided, there are two maximum cliques as shown, each with the depicted set of Vertices.

There can be Cliques which are smaller in size (i.e. containing lesser number of Vertcies), but no clique is strictly greater than the ones depicted.

```python
%%manim -qm -v WARNING Scene_3_MCP
# Chat GPT
class Scene_3_MCP(Scene):
    def construct(self):
        # --- Title on top ---
        title = Text("The Maximum Clique Problem (MCP)", font_size=34, color=YELLOW)
        title.to_edge(UP, buff=0.5)
        self.play(Write(title))
        self.wait(1)

        # --- Definition text on the right ---
        definition_title = Text("Definition:", font_size=28, color=BLUE, weight=BOLD)
        definition_title.to_edge(RIGHT, buff=3.5).shift(UP * 2)

        definition_text = Text(
            "MCP = Find the largest Clique\nin a Graph.\n\n"
            "- Output: vertices forming\n  the largest clique\n"
            "- A graph may have more than\n  one maximum clique",
            font_size=24,
            line_spacing=1.2
        )
        definition_text.next_to(definition_title, DOWN, buff=0.5, aligned_edge=LEFT)

        self.play(Write(definition_title))
        self.wait(0.5)
        self.play(Write(definition_text))
        self.wait(2)

        # --- Graph on left ---
        vertices = {
            'A': UP * 2 + LEFT * 3,
            'B': UP * 1 + LEFT * 4,
            'C': DOWN * 0.5 + LEFT * 3.5,
            'D': UP * 1.5 + LEFT * 1.5,
            'E': DOWN * 0.5 + LEFT * 1.5,
            'F': DOWN * 2 + LEFT * 3,
        }

        vertex_objects = {}
        for label, pos in vertices.items():
            circle = Circle(radius=0.3, fill_color=WHITE, fill_opacity=1, stroke_color=BLACK)
            circle.move_to(pos)
            text = Text(label, font_size=24, color=BLACK).move_to(pos)
            vertex_objects[label] = VGroup(circle, text)

        edges = [
            ('A', 'B'), ('A', 'D'), ('A', 'E'), ('A', 'F'),
            ('B', 'C'), ('B', 'D'),
            ('C', 'F'),
            ('D', 'E'), ('D', 'F'),
            ('E', 'F')
        ]

        edge_objects = [Line(vertices[s], vertices[t], color=GRAY, stroke_width=3) for s, t in edges]
        graph = VGroup(*edge_objects, *vertex_objects.values())
        graph.shift(LEFT * 1.0)

        self.play(Create(VGroup(*edge_objects)), run_time=2)
        self.play(FadeIn(VGroup(*vertex_objects.values())), run_time=1.5)
        self.wait(1)

        # --- Highlight first maximum clique (A, D, E, F) ---
        clique1_vertices = ['A', 'D', 'E', 'F']
        clique1_edges = [('A', 'D'), ('A', 'E'), ('A', 'F'), ('D', 'E'), ('D', 'F'), ('E', 'F')]
        clique1 = VGroup(
            *[Circle(radius=0.4, color=RED, stroke_width=4, fill_opacity=0).move_to(vertices[v]) for v in clique1_vertices],
            *[Line(vertices[s], vertices[t], color=RED, stroke_width=6) for s, t in clique1_edges]
        )
        clique1_text = Text("Maximum Clique 1: {A, D, E, F}", font_size=22, color=RED)
        clique1_text.next_to(definition_text, DOWN, buff=1)

        self.play(Create(clique1), Write(clique1_text))
        self.wait(2)

        # --- Highlight second maximum clique (B, A, D) if it’s same size ---
        clique2_vertices = ['A', 'B', 'D']
        clique2_edges = [('A', 'B'), ('A', 'D'), ('B', 'D')]
        clique2 = VGroup(
            *[Circle(radius=0.4, color=ORANGE, stroke_width=4, fill_opacity=0).move_to(vertices[v]) for v in clique2_vertices],
            *[Line(vertices[s], vertices[t], color=ORANGE, stroke_width=6) for s, t in clique2_edges]
        )
        clique2_text = Text("Maximum Clique 2: {A, B, D}", font_size=22, color=ORANGE)
        clique2_text.next_to(clique1_text, DOWN, buff=0.3)

        self.play(Create(clique2), Write(clique2_text))
        self.wait(2)

        # --- Wrap up ---
        summary = Text("No clique is larger than these.", font_size=24, color=GREEN)
        summary.next_to(clique2_text, DOWN, buff=0.5)
        self.play(Write(summary))
        self.wait(3)
```
